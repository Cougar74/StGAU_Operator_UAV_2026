# Поиск в неупорядоченных массивах

```python
import numpy as np
```

1. Начало: Получаем на вход массив и искомое значение
1. Итерация: Последовательно перебираем элементы массива, начиная с первого
1. Сравнение: Для каждого элемента проверяем, равен ли он искомому значению
1. Результат:
- Если найден совпадающий элемент – возвращаем его позицию (индекс)
- Если весь массив просмотрен и элемент не найден – возвращаем специальное значение (обычно -1 или None)

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
```

```python
numbers = [1, 3, 5, 8, 9, 12]

print(linear_search(numbers, 9))
print(linear_search(numbers, 55))
```

```python
%timeit linear_search(numbers, 8)
```

# Дихотомический поиск

1. Находим средний элемент массива
1. Сравниваем его с искомым значением
1. Если средний элемент равен искомому – поиск завершен
1. Если искомое значение меньше среднего элемента – продолжаем поиск в левой половине
1. Если искомое значение больше среднего элемента – продолжаем поиск в правой половине
1. Повторяем процесс, пока не найдем элемент или не убедимся в его отсутствии

```python
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = left + (right - left) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

```python
numbers = [1, 3, 5, 8, 9, 12]

print(binary_search(numbers, 9))
print(binary_search(numbers, 55))
```

```python
N = 1_000
array = np.random.choice(N, size=N, replace=False)
array = np.sort(array)

%timeit binary_search(array, N-1)
```

# Алгоритмы сортировки

## Пузырьковая сортировка (Bubble Sort)

Алгоритм работает по следующей схеме:

1. Проходим по массиву от начала до конца
1. Сравниваем каждую пару соседних элементов
1. Если элементы расположены в неправильном порядке (предыдущий больше следующего), меняем их местами
1. Повторяем процесс до тех пор, пока массив не будет полностью отсортирован

```python
arr = [64, 34, 25, 12, 22, 11, 90]
```

```python
def buble_sort(arr):
    for i in range(len(arr) - 1):
        # print(f'\nПроход {i + 1}:')
        # print(f'До прохода: {arr}')
        
        for j in range(len(arr) - 1):
            # print(f'  Сравниваем {arr[j] = } и {arr[j+1] = }', end='')
            
            if arr[j] < arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                # print(f' -> МЕНЯЕМ местами')
                
            # else:
            #     # print(f' -> не меняем')
        
        # print(f'После прохода: {arr}')
        
    return arr
```

```python
buble_sort(arr.copy())
```

```python
N = 1_000
array = np.random.choice(N, size=N, replace=False)

%timeit buble_sort(array.copy())
```

```python
N = 10_000
array = np.random.choice(N, size=N, replace=False)

%timeit buble_sort(array.copy())
```

# Сортировка слиянием (Merge Sort)

Работает по принципу рекурсивного алгоритма

1. Базовый случай: Если массив содержит 0 или 1 элемент, он уже отсортирован
2. Разделение: Находим середину массива: `mid = len(arr) // 2`
3. Рекурсивный вызов:
- Сортируем левую половину: `left = merge_sort(arr[:mid])`
- Сортируем правую половину: `right = merge_sort(arr[mid:])`
4. Слияние: Объединяем отсортированные половины в один массив

```python
def merge_sort(arr, level = 0):
    print(f'#{level}: {arr}')
    
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    
    left_half = merge_sort(arr[:mid], level + 1)
    right_half = merge_sort(arr[mid:], level + 1)
    
    return merge(left_half, right_half, level)

def merge(left, right, level):
    print(f'  Слияние на уровне #{level}')
    print(f'  {left} -- {right}')
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        print(f'    {i}: {left[i]} <= {right[j]}: {j}')
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    
    print(f'  {result}')
    return result
```

```python
def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)
```

```python
arr = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(arr))
```

```python
import os
```

```python
PATH_DIR = 'routes/'
FILES = os.listdir(PATH_DIR)
```

```python
def parse_file(file_name):
    with open(file_name, 'r') as f:
        data = f.readlines()
        
    data = data[1:]
    
    d = []
    for line in data:
        d.append(line.strip().split(','))
        
    # -> 
    # d = [line.strip().split(',') for line in data]
    
    d.sort(key=lambda x: x[0])
    
    formated_data = []
    for line in d:
        new_line = [float(v) if i else v for i, v in enumerate(line)]
        formated_data.append(new_line)
    
    return formated_data
```

```python
parse_file(PATH_DIR + FILES[0])
```

```python
def check_line(data):
    route_id, distance_km, speed_kmh, charge_pct, consumption_pct_per_km, reserve_pct = data
    
    if charge_pct == 'inf' or distance_km == 'inf' or consumption_pct_per_km == 'inf' or reserve_pct == 'inf':
        return False, 'inf'

    if speed_kmh <= 0:
        return False, 'скорость должна быть больше 0'
    if distance_km <= 0:
        return False, 'расстояние должно быть больше 0'
    if charge_pct <= 0 or charge_pct > 100:
        return False, 'заряд должен быть больше 0 и меньше 100'
    if consumption_pct_per_km <= 0 or consumption_pct_per_km > 100:
        return False, 'потребление должно быть больше 0 и меньше 100'
    if reserve_pct <= 0 or reserve_pct > 100:
        return False, 'резерв должен быть больше 0 и меньше 100'
    
    time = distance_km / speed_kmh * 60
    consumption = distance_km * consumption_pct_per_km
    charge = charge_pct - consumption
    
    return True if charge >= reserve_pct else False, None
```

```python
for file in FILES:
    print(f'{file}')
    
    data = parse_file(PATH_DIR + file)
    errors = 0
    failed = 0
    passed = 0
    dublicates = []
    prev_id = None
    
    for line in data:
        result, error = check_line(line)
        
        if line[0] == prev_id:
            dublicates.append(line[0])
        prev_id = line[0]
        
        if error:
            print(f'{line[0]}: {error}')
            errors += 1
        
        if not result:
            print(f'{line[0]}: Не прошел проверку')
            failed += 1
        else:
            print(f'{line[0]}: OK')
            passed += 1
    
    print(f'Повторы: {dublicates}')
    print(f'Прошло проверку: {passed}')
    print(f'Ошибки: {errors}')
    print(f'Не прошло: {failed}')
    print('--------------------------------')
```

## Основные принципы ООП
### Наследование

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.speed = 0
        
    def get_descriptive_name(self):
        return f"{self.year} {self.make} {self.model}"

    def get_speed(self):
        return f"Машина едет со скоростью {self.speed} км/ч"
    
    def update_speed(self, new_value):
        if new_value < 0:
            print("Скорость не может быть отрицательной")
            return
        
        self.speed = new_value
        print(f"Новая скорость: {self.speed} км/ч")
```

```python
class ElectricCar(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery = 100
    
    def get_battery(self):
        return f"Заряд батареи: {self.battery}%"
    
    def update_battery(self, new_value):
        if not 0 <= new_value <= 100:
            print("Заряд должен быть от 0 до 100%")
            return
        
        self.battery = new_value
        return f"Новый заряд батареи: {self.battery}%"
```

```python
my_tesla = ElectricCar('Tesla', 'Model S', 2019)

print(my_tesla.get_descriptive_name())

print(my_tesla.get_speed())
my_tesla.update_speed(60)
print(my_tesla.get_speed())

print(my_tesla.get_battery())
my_tesla.update_battery(65)
print(my_tesla.get_battery())
```

### Инкапсуляция

```python
class BankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Начальный баланс не может быть отрицательным")
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Сумма пополнения должна быть положительной")
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0 or amount > self.__balance:
            raise ValueError("Недопустимая сумма списания")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance
```

```python
account = BankAccount(1000)
print(account.get_balance()) 
account.deposit(500)
print(account.get_balance())
account.withdraw(200)
print(account.get_balance())
```

### Абстракция

```python
from abc import ABC, abstractmethod
import math 
```

```python
class Shape(ABC):
    # Декоратор
    
    @abstractmethod
    def area(self):
        pass
```

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2
```

```python
circle = Circle(6)
print(f"Площадь круга: {circle.area():.2f}")
```

```python
shape = Shape()
```

```python
class Car(ABC):
    @abstractmethod
    def forward(self):
        pass
    
    @abstractmethod
    def back(self):
        pass
    
    @abstractmethod
    def turn_left(self):
        pass
    
    @abstractmethod
    def turn_right(self):
        pass
```

```python
class Train(Car):
    def forward(self):
        print('Поезд едет вперед')
        
    def back(self):
        print('Поезд едет назад')
    
    def turn_left(self):
        print('Поезд не умеет поворачивать!')
        
    def turn_right(self):
        print('Поезд не умеет поворачивать!')
```

```python
train = Train()
```
